```python
from abc import ABC, abstractmethod

class Absclass(ABC):
  def print(self, x):
    print("Value:", x)

  @abstractmethod
  def task(self):
    print("We are in the task of Absclass")

class test_class(Absclass):
  def task(self):
    print("We are in the task of test_class")

obj = test_class()
obj.task()
obj.print(100)
```

```python
from abc import ABC

class Absclass(ABC):

  def move(self):
    pass

class Human(Absclass):
  def move(self):
    print("I can walk and run")

class Snake(Absclass):
  def move(self):
    print("I can crawl")

class Dog(Absclass):
  def move(self):
    print("I can bark")

class Lion(Absclass):
  def move(self):
    print("I can roar")

H = Human()
H.move()

S = Snake()
S.move()

D = Dog()
D.move()

L = Lion()
L.move()

```
