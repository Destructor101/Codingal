```python
class BMW:
  def __init__(self):
    self.name = "BMW"

  def fuel_type(self):
    print("Petrol or Electric")

  def max_speed(self):
    print("250km/h")

class Ferrari:
  def __init__(self):
    self.name = "Ferrari"

  def fuel_type(self):
    print("Petrol")

  def max_speed(self):
    print("360km/h")

obj_BMW = BMW()
obj_Ferrari = Ferrari()

for car in [obj_BMW, obj_Ferrari]:
  print(car.name)
  car.fuel_type()
  car.max_speed()
  print()

```
